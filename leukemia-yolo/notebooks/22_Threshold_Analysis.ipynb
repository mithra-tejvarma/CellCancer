# Validation threshold study

The objective is maximum validation specificity subject to blast sensitivity at least 0.95. Scores are uncalibrated.

Run from the existing `leukemia-yolo` Conda kernel. Source datasets and earlier results are read only.

```python
from pathlib import Path
import sys
import pandas as pd
import matplotlib.pyplot as plt
ROOT = Path.cwd().resolve()
if ROOT.name != 'leukemia-yolo': ROOT = ROOT.parent
assert (ROOT / 'config.local.yaml').is_file()
print(sys.executable, ROOT)
```

## Validation-only threshold candidates

No threshold is chosen from fold_2.

```python
import json
p = ROOT/'outputs/classifier/improvement_audit'
study = json.loads((p/'validation_threshold_study.json').read_text())
print(study['objective']); display(pd.DataFrame([study['selected'], study['macro_f1_best']]))
```

```python
curve = pd.read_csv(p/'validation_threshold_curve.csv')
ax=curve.plot(x='threshold',y=['blast_sensitivity','normal_specificity','macro_f1'],figsize=(9,4)); ax.axhline(.95,color='gray',ls=':'); plt.show()
```

```python
comparison=pd.read_csv(p/'validation_model_comparison.csv')
display(comparison[['model','policy','threshold','blast_sensitivity','normal_specificity','macro_f1','roc_auc']])
print('ROC-AUC does not depend on a decision threshold.')
```

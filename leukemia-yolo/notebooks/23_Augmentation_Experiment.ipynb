# Mild augmentation experiment

Controlled continuation from the verified full checkpoint. Compare with the unchanged-policy control initialized identically.

Run from the existing `leukemia-yolo` Conda kernel. Source datasets and earlier results are read only.

```python
from pathlib import Path
import sys
import pandas as pd
import matplotlib.pyplot as plt
ROOT = Path.cwd().resolve()
if ROOT.name != 'leukemia-yolo': ROOT = ROOT.parent
assert (ROOT / 'config.local.yaml').is_file()
print(sys.executable, ROOT)
```

## Launch in a Windows-safe subprocess

Set `RUN_TRAINING=True` only when ready for a new run. The process streams its progress and writes a unique output directory.

```python
import subprocess
RUN_TRAINING=False
command=[sys.executable,'-u',str(ROOT/'scripts/run_improvement_trial.py'),'--arm','augmentation_trial','--epochs','8','--patience','3','--batch-size','4','--accumulation','2','--workers','2','--lr','0.00001']
print(' '.join(command))
if RUN_TRAINING: subprocess.run(command,cwd=ROOT,check=True)
```

```python
base=ROOT/'outputs/classifier/improvements/augmentation_trial'
runs=sorted(base.glob('trial_*/completed.json'))
if runs:
    run=runs[-1].parent
    print(run)
    history=pd.read_csv(run/'training_history.csv')
    display(history[['epoch','val_specificity','val_sensitivity','val_macro_f1','val_roc_auc']])
    history.plot(x='epoch',y=['val_specificity','val_sensitivity','val_macro_f1']);plt.show()
else: print('Trial pending')
```

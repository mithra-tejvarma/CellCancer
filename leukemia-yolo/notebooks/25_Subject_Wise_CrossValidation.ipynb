# Subject-wise cross-validation

Optional internal cross-validation. Each fold must initialize independently from ImageNet weights.

Run from the existing `leukemia-yolo` Conda kernel. Source datasets and earlier results are read only.

```python
from pathlib import Path
import sys
import pandas as pd
import matplotlib.pyplot as plt
ROOT = Path.cwd().resolve()
if ROOT.name != 'leukemia-yolo': ROOT = ROOT.parent
assert (ROOT / 'config.local.yaml').is_file()
print(sys.executable, ROOT)
```

## Verified plan

The plan excludes fold_2 and checks subject, path, and image-hash separation.

```python
import subprocess, json
command=[sys.executable,str(ROOT/'scripts/subject_cv.py'),'--folds','5']
subprocess.run(command,cwd=ROOT,check=True)
plans=sorted((ROOT/'outputs/classifier/improvements/cross_validation').glob('cv_*/cv_plan.json'))
plan=json.loads(plans[-1].read_text());display(pd.DataFrame(plan['folds'])[['fold','train_cells','val_cells','val_normal_cells','val_blast_cells']])
```

```python
RUN_CV=False
if RUN_CV: subprocess.run([sys.executable,'-u',str(ROOT/'scripts/subject_cv.py'),'--run','--folds','5','--epochs','8'],cwd=ROOT,check=True)
else: print('Five independent fold runs are optional and have not started.')
```

```python
summaries=sorted((ROOT/'outputs/classifier/improvements/cross_validation').glob('cv_*/cv_summary.json'))
if summaries:
    result=json.loads(summaries[-1].read_text());display(pd.DataFrame(result['folds']))
else: print('No completed cross-validation metrics to report.')
```
